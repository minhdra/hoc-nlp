Chạy thử PhoBERT: biến một câu tiếng Việt thành vector. Lần đầu sẽ tải model (~500MB).

```python
from transformers import AutoTokenizer, AutoModel
tok = AutoTokenizer.from_pretrained('vinai/phobert-base')
model = AutoModel.from_pretrained('vinai/phobert-base').to('cuda')

# PhoBERT cần câu đã tách từ (từ ghép nối bằng _)
cau = 'Tôi đang học xử_lý ngôn_ngữ tự_nhiên'
x = tok(cau, return_tensors='pt').to('cuda')
with torch.no_grad():
    out = model(**x)
print(x['input_ids'])
print(out.last_hidden_state.shape)  # (1, số_token, 768)
```
